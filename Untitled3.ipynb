```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

print("Libraries loaded successfully!")
```

```python
url = "https://raw.githubusercontent.com/abcom-mltutorials/mall/main/Mall_Customers.csv"

df = pd.read_csv(url)

print(df.head())
print("Dataset shape:", df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nStatistical summary:")
print(df.describe())
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Age", kde=True)
plt.title("Distribution of Customer Age")
plt.xlabel("Age (years)")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    hue="Gender"
)
plt.title("Annual Income vs Spending Score")
plt.xlabel("Annual Income (thousand dollars)")
plt.ylabel("Spending Score")
plt.tight_layout()
plt.show()
```

```python
features = [
    "Annual Income (k$)",
    "Spending Score (1-100)"
]

X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Feature scaling completed.")
```

```python
inertias = []

for k in range(2, 10):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    model.fit(X_scaled)
    inertias.append(model.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(2, 10), inertias, marker="o")
plt.title("Elbow Method for Choosing K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.xticks(range(2, 10))
plt.grid(True)
plt.show()
```

```python
for k in range(2, 10):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = model.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)

    print(f"K = {k}, Silhouette Score = {score:.3f}")
```

```python
k = 5

kmeans = KMeans(
    n_clusters=k,
    random_state=42,
    n_init=10
)

df["Cluster"] = kmeans.fit_predict(X_scaled)

print("Clustering completed!")
print(df[["CustomerID", "Cluster"]].head())
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Annual Income (k$)",
    y="Spending Score (1-100)",
    hue="Cluster",
    palette="tab10",
    s=80
)

plt.title("Customer Segmentation Using K-Means")
plt.xlabel("Annual Income (thousand dollars)")
plt.ylabel("Spending Score (1-100)")
plt.legend(title="Cluster")
plt.tight_layout()
plt.show()
```

```python
summary = df.groupby("Cluster").agg(
    Customer_Count=("Cluster", "size"),
    Average_Age=("Age", "mean"),
    Average_Income=("Annual Income (k$)", "mean"),
    Average_Spending=("Spending Score (1-100)", "mean")
).round(2)

print(summary)
```

```python
df.to_csv("clustered_customers.csv", index=False)
summary.to_csv("cluster_summary.csv")

print("Files saved successfully!")
```
